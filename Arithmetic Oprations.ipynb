```python
import numpy as np

A = np.array([[1, 2, 1], [2, 2, 3]])
B = np.array([3, 4, 5])

print(A)
print(B)
```

```python
# Addition
print(np.add(A, B))

# Subtraction
print(np.subtract(B, A))

# Multiplication
print(np.multiply(A, B))

# Division
print(np.divide(A, B))
```

```python
# Power operations
print(np.power(A, 2))

print(np.power(A, B))
```
