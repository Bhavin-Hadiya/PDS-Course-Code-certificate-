```python
import numpy as np
arr = np.array([[4, 3, 2],
                [10, 1, 0],
                [5, 8, 24]])
print(arr)
```

```python
# Minimum values
print(np.amin(arr))
print(np.amin(arr, axis=0))
print(np.amin(arr, axis=1))
```

```python
# Maximum values
print(np.max(arr))
print(np.amax(arr, axis=0))
print(np.amax(arr, axis=1))
```

```python
# Median, Mean, Standard Deviation
print(np.median(arr))
print(np.mean(arr))
print(np.std(arr))
```

```python
# Trigonometric functions with different angles
deg = np.array([15, 45, 75, 120])

print(np.sin(deg*np.pi/180))

print(np.cos(deg*np.pi/180))

print(np.tan(deg*np.pi/180))
```
