```python
import pandas as pd

l = [x*2 for x in range(5)]

s = pd.Series(l)
```

```python
# Access by numeric index
print(s[2])
```

```python
# Custom string indices
s = pd.Series(l, index=['p','q','r','s','t'])

print(s['t'])
```

```python
# Duplicate indices
s = pd.Series(l, index=['x','y','z','z','y'])
print(s)

print(s['z'])
```

```python
# Dictionary with changed values
data = {'x': 10, 'y': 20, 'z': 30, 'w': 40}

# Create Series directly from dictionary
s = pd.Series(data)
print(s)
```

```python
# Restrict Series to specific keys
s = pd.Series(data, index=['x','y'])
print(s)
```

```python
# Include an index not present in dictionary
s = pd.Series(data, index=['x','y','z','q'])
print(s)
```

```python
# Create a small DataFrame (changed values)
df = pd.DataFrame({
    'Name': ['Aarav', 'Meera', 'Kunal', 'Priya', 'Rohan'],
    'Age': [22, 27, 35, 19, 40],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male']
})

print(df.head())
```

```python
# Write DataFrame to JSON file
df.to_json('New_File.json')

# Read JSON file back into DataFrame
df_json = pd.read_json('New_File.json')
print(df_json)
```
